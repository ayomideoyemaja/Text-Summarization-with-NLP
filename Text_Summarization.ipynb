```python
import nltk
import string
from heapq import nlargest

# Ensure NLTK dependencies are downloaded
nltk.download('stopwords')
nltk.download('punkt_tab')

```

```python
text = """Network Security is an essential aspect of our daily lives nowadays.
Classical protection techniques, such as rule-based firewalls and signature-based detection,
are limited in many ways. Those methods may need assistance efficiently responding to the
continuously increasing demand for protection and the constantly evolving attacking strategies.
IDS, applied to network security, helps us maintain, access, and protect our vulnerable information.
This paper suggests a machine learning model to efficiently reduce the informational structure of the network
and detect the attacks prominently. The Gain Ratio method reduces the data dimension, and Random Forest is
applied for classification purposes. We test and train the model with the NSL-KDD data set, and we
provide proof that the implementation of the Gain Ratio can help the IDS to successfully recognize
malicious traffic by evaluating the reduced structure of the network and detecting the attack."""

# Determine the length of the summary
if text.count(". ") > 20:
    length = int(round(text.count(". ") / 10, 0))
else:
    length = 1

# Remove punctuation
nopunc = "".join([char for char in text if char not in string.punctuation])

# Tokenize words and remove stopwords
stop_words = set(nltk.corpus.stopwords.words('english'))
processed_text = [word.lower() for word in nopunc.split() if word.lower() not in stop_words]

# Calculate word frequencies
word_freq = {}
for word in processed_text:
    word_freq[word] = word_freq.get(word, 0) + 1

# Normalize word frequencies
max_freq = max(word_freq.values())
for word in word_freq:
    word_freq[word] /= max_freq

# Tokenize sentences and calculate sentence scores
sent_list = nltk.sent_tokenize(text)
sent_score = {}
for sent in sent_list:
    for word in nltk.word_tokenize(sent.lower()):
        if word in word_freq:
            sent_score[sent] = sent_score.get(sent, 0) + word_freq[word]

# Extract the top sentences
summary_sents = nlargest(length, sent_score, key=sent_score.get)
summary = " ".join(summary_sents)

# Print the summary
print("Summary:")
print(summary)
```
